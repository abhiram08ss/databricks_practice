```python
%sql
CREATE SCHEMA IF NOT EXISTS my_catalog.demo_schema;
```

```python
data = [
    (1, "John", 25, "Hyderabad", 50000),
    (2, "Priya", 28, "Bangalore", 65000),
    (3, "Rahul", 22, "Chennai", 45000),
    (4, "Sarah", 31, "Mumbai", 75000),
    (5, "David", 26, "Delhi", 55000)
]

columns = ["id", "name", "age", "city", "salary"]

df = spark.createDataFrame(data, columns)

df.show()
df.printSchema()
```

```python
df.write\
    .format("delta")\
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
        .saveAsTable("my_catalog.demo_schema.users",path="abfss://abhiramdirectroy20260817@abhiramblobstorage.dfs.core.windows.net/databricks/replica-test")
```
