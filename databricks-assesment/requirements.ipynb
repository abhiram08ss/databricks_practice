```python
path = 'abfss://databricksminiproject@abhiramblobstorage.dfs.core.windows.net'
catalog = 'databricks_assesment'
```

```python
def create_schema(catalog, schema, storage_path):
    spark.sql(f"""
        CREATE SCHEMA IF NOT EXISTS {catalog}.{schema}
        MANAGED LOCATION '{storage_path}/{schema}'
    """)
```

```python
def create_volume(catalog, schema, volume):
    spark.sql(f"""
        CREATE VOLUME IF NOT EXISTS {catalog}.{schema}.{volume}
    """)
```

```python

create_schema(catalog,'landing',path)
```

```python
create_schema(catalog,'bronze',path)
create_schema(catalog,'silver',path)
create_schema(catalog,'gold',path)
```
